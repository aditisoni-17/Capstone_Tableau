## 1.6 Missing Values Summary

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
missing_df[missing_df['Missing Count'] > 0].sort_values('Missing Count', ascending=False)
```

```python
print(f'Total rows: {len(df):,}')
print(f'Rows with missing CustomerID: {df["CustomerID"].isnull().sum():,} ({df["CustomerID"].isnull().mean()*100:.1f}%)')
print(f'Rows with missing Description: {df["Description"].isnull().sum():,} ({df["Description"].isnull().mean()*100:.1f}%)')
```

## 1.7 Initial Observations

1. **Size:** 541,909 rows × 8 columns — large transactional dataset.
2. **Missing CustomerID:** ~135,080 rows (24.9%) have no CustomerID — these must be dropped for customer-level analysis.
3. **Missing Description:** 1,454 rows (~0.3%) — minor, can be handled during cleaning.
4. **InvoiceDate:** Stored as string — needs datetime conversion.
5. **Quantity:** Contains negative values — likely cancellations (invoices starting with 'C').
6. **UnitPrice:** Contains zero and negative values — needs filtering.
7. **Country:** Predominantly United Kingdom — international sales exist but are a small fraction.
8. **Data Period:** December 2010 to December 2011 (approximately 1 year).
